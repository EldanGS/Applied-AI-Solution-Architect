<details><summary>Exercise 1 — Qualify and rank use cases: reference solution</summary>

```python
def qualify(uc: dict, threshold: float = 3.5) -> dict:
    failed = [g for g in GATES if not uc.get(g, False)]
    if failed:
        return {"decision": "park", "failed_gates": failed, "value": 0.0, "feasibility": 0.0}
    value = round(sum(w * uc.get(k, 0) for k, w in VALUE_WEIGHTS.items()), 2)
    feas = round(sum(w * uc.get(k, 0) for k, w in FEAS_WEIGHTS.items()), 2)
    if value >= threshold and feas >= threshold:
        decision = "quick_win"
    elif value >= threshold:
        decision = "strategic_bet"
    elif feas >= threshold:
        decision = "self_service"
    else:
        decision = "park"
    return {"decision": decision, "failed_gates": [], "value": value, "feasibility": feas}


def rank(use_cases: list[dict], threshold: float = 3.5) -> list[str]:
    def key(uc):
        q = qualify(uc, threshold)
        return (ORDER.index(q["decision"]), -q["value"], uc["name"])
    return [uc["name"] for uc in sorted(use_cases, key=key)]
```

<details><summary>Exercise 2 — An honest value case and its most sensitive input: reference solution</summary>

```python
def value_case(x: dict) -> dict:
    used = x["volume"] * x["adoption"]
    gross = used * (x["min_before"] - x["min_after"]) / 60 * x["rate"]
    hard, soft = gross * x["hard_share"], gross * (1 - x["hard_share"])
    review = used * x["review_share"] * x["review_minutes"] / 60 * x["rate"]
    yearly = used * x["run_cost_per_task"] + review + x["fixed_cost"]
    net = hard - yearly
    total = yearly + x["build_cost"]
    return {"hard_benefit": round(hard), "soft_benefit": round(soft), "yearly_cost": round(yearly),
            "roi_year1": round((net - x["build_cost"]) / total, 2),
            "payback_months": round(x["build_cost"] / (net / 12), 1) if net > 0 else None}


def most_sensitive(base: dict, keys: list[str], delta: float = 0.2) -> str:
    def swing(k):
        lo = value_case({**base, k: base[k] * (1 - delta)})["roi_year1"]
        hi = value_case({**base, k: base[k] * (1 + delta)})["roi_year1"]
        return abs(hi - lo)
    return max(keys, key=swing)      # max() keeps the first key on a tie
```

<details><summary>Exercise 3 — Trace business, functional and infrastructure requirements: reference solution</summary>

```python
def trace_gaps(business: dict, functional: dict, infra: dict, required_categories: set[str]) -> dict:
    covered = {br for fr in functional.values() for br in fr["traces_to"] if br in business}
    dangling = [f"{fid}->{ref}" for fid, fr in functional.items()
                for ref in fr["traces_to"] + fr["needs"] if ref not in business and ref not in infra]
    return {
        "br_without_fr": sorted(set(business) - covered),
        "fr_without_ir": sorted(fid for fid, fr in functional.items() if not any(n in infra for n in fr["needs"])),
        "dangling_refs": sorted(dangling),
        "missing_categories": sorted(set(required_categories) - {ir["category"] for ir in infra.values()}),
    }
```

<details><summary>Exercise 4 — Go, iterate or no-go: reference solution</summary>

```python
def go_no_go(results: dict, criteria: list[dict], min_items: int = 200) -> dict:
    ops = {"<=": lambda v, t: v <= t, ">=": lambda v, t: v >= t, "==": lambda v, t: v == t}
    not_measured = sorted(c["metric"] for c in criteria if results.get(c["metric"]) is None)
    failed = {c["metric"] for c in criteria
              if c["metric"] in not_measured or not ops[c["op"]](results[c["metric"]], c["threshold"])}
    guardrail_failed = any(c["kind"] == "guardrail" and c["metric"] in failed for c in criteria)
    decision = "no-go" if guardrail_failed else ("iterate" if failed else "go")
    if results.get("n_eval_items", 0) < min_items:
        failed.add("n_eval_items")
        if decision == "go":
            decision = "iterate"
    return {"decision": decision, "failed": sorted(failed), "not_measured": not_measured}
```
